## Exercise 2: Order of Operations

Each expression needed the parentheses moved so that the addition happened before the multiplication, or the other way round.

* `2 + 8 * 2` multiplies first, giving 18. Wrapping the addition in parentheses forces it to happen first.
* `4 * 3 + 2` multiplies first, giving 14. Same fix.
* `3 * (5 + 5)` already has parentheses, and they are the problem. Removing them lets the multiplication happen first, giving `3 * 5 + 5`.

```python
print((2 + 8) * 2)
print(4 * (3 + 2))
print(3 * 5 + 5)
```

## Exercise 3: Tracing Assignment

The output is `6`, then `5`, then `6`.

`b = a` copies the value that `a` refers to at that moment, which is 5. Changing `a` afterwards does nothing to `b`. This is the difference between assignment and equality again: `b` was not defined to be "the same as `a` forever", it was handed a value once.

The third answer catches people out. `count += count` means `count = count + count`, so the right-hand side is worked out first using the current value, 3, giving 6.

```python
a = 5
b = a
a = a + 1

print(a)  # 6, because a was reassigned on the line above
print(b)  # 5, because b was given a's value before that happened

count = 0
count += 3      # count is now 3
count += count  # count = count + count, so 3 + 3

print(count)    # 6
```

## Exercise 4: Using Variables

The formula translates directly into Python. The only real decisions are what to call the result, and whether to use parentheses.

The parentheses around the subtraction are essential here. Without them, the division would happen first and the answer would be wrong.

```python
initial_mass_g = 78.4
final_mass_g = 65.1

percentage_mass_change = (final_mass_g - initial_mass_g) / initial_mass_g * 100

print(percentage_mass_change)
```

The result is `-16.964285714285726`. The sign is negative because the sample lost mass, and the long string of decimal places is simply how Python reports the result of a division. You will see how to round it in a later session.

## Extra Exercise 1: Whitespace

Nothing about the calculations changed. Only the spacing did, and that's the difference between code you can skim and code you have to decode.

Look closely at the last line. `2 * 2 - -4` is subtracting negative four, so the two minus signs are doing different jobs: the first is subtraction, the second makes the 4 negative. Spacing them differently makes that visible.

```python
print((10 + 2 + 2) / 10 * 5)    # evaluates to 7.0
print(100 / 10 / (2 / 2))       # evaluates to 10.0
print(2 * 2 - -4)               # evaluates to 8
```

## Extra Exercise 3: A Calculation of Your Own

There is no single right answer here, but a good solution will have one step per line, names that state what each value is and what units it is in, and a comment on anything that would not be obvious to a reader from outside your field.

Below is an example: converting a spectrophotometer reading into a concentration using the Beer-Lambert law.

```python
# Absorbance measured by the spectrophotometer (no units)
absorbance = 0.482

# Molar absorption coefficient for this compound, in L per mol per cm
molar_absorptivity_l_mol_cm = 6300

# Path length of the cuvette, in cm
path_length_cm = 1.0

# Beer-Lambert law: concentration = absorbance / (absorptivity * path length)
concentration_mol_l = absorbance / (molar_absorptivity_l_mol_cm * path_length_cm)

print(concentration_mol_l)
```
