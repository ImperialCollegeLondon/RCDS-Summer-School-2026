### Exercise 3: FizzBuzz
In the cell below, write an if block that prints "Fizz" if ```a``` is a multiple of 3 and "Buzz" if it's a multiple of 5. If it's a multiple of both 3 and 5, print "FizzBuzz" instead. If it's not a multiple of 3 or 5, print the value of ```a``` instead. Test your code for a few values of ```a```.

The modulo operator may be useful to you. Carefully consider the order of boolean expressions.

````python
a = 1

# The statements are checked in order, so we need to check the FizzBuzz case first.
# This is because if the FizzBuzz case is met, so are the Fizz and Buzz cases.
if a % 3 == 0 and a % 5 == 0:
  # Print "FizzBuzz" if a is a multiple of 3 and 5.
  # If this if statement isn't met, we know a isn't a multiple of 3 and/or it's not a multiple of 5.
  print("FizzBuzz")
elif a % 3==0:
  # Print "Fizz" if a is a multiple of 3 and not 5.
  print("Fizz")
elif a % 5==0:
  # Print "Buzz" if a is a multiple of 5 and not 3.
  print("Buzz")
else:
  # If the else is reached, a is not a multiple of 3 or 5.
  print(a)
````
