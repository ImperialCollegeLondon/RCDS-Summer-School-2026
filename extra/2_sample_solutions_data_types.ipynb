### Exercise 2

Your task is to find errors in the following code. Carefully read the output of python interpreter to pinpoint them.

The desired output of the code below:
<pre>
=== CHARACTER PROFILE ===
Hero Name: Arthur Pendragon
Save ID: arthur_pendragon
Health Points: 100 HP
Magic Ability: True
Top Speed: 7.5 m/s
=========================
</pre>


The correct code:

````python
# Raw profile data entered by the player
raw_name = "   aRtHuR PeNdRaGoN   "
health_points = 100
has_magic = True
top_speed = 7.5


# Strip whitespace and use title case for names
clean_name = raw_name.strip().title()

# Make it lowercase and swap spaces for underscores
save_id = clean_name.lower().replace(" ", "_")

# Output the final profile layout
print("=== CHARACTER PROFILE ===")
print("Hero Name:", clean_name)
print("Save ID:", save_id)
print("Health Points:", str(health_points) + " HP")
print("Magic Ability:", has_magic)
print("Top Speed:", str(top_speed) + " m/s")
print("=========================")
````
