### Exercise 1

Write a function ```evaluate_linear_equation``` that uses linear equation $y = mx + c$ to calculate value of $y$ given value of $x$.

Your function will have three arguments:
- slope $m$
- y-intercept $c$
- value of $x$

````python
# Write your function here
def evaluate_linear_equation(m, c, x):
    y = m * x + c
    return y


# Sample code that should print the value 11
print(evaluate_linear_equation(2, 3, 4))  

# Sample code that should print the value 19
print(evaluate_linear_equation(2, 3, 8))  

# Sample code that should print the value 3
print(evaluate_linear_equation(0, 3, 2))  

# Sample code that should print the value 10
print(evaluate_linear_equation(2, 0, 5))  
````
