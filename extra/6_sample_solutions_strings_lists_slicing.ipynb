## Adding Spaces

The function should receive one argument - a string to be operated on. We can use the accumulator pattern to build a new string with spaces between each character, so the first thing we need to do is initialise an empty string to hold the result. We can then loop a range created using the length of the input string. This means that the loop will iterate over the indices of the string, which we can use to access each character. Within the loop, we add each character to the result string. We also check the current index to see if it is the last character in the string. If it is not, we add a space after the character. Finally, after the loop completes, we return the result string.

```python
import ipytest
ipytest.autoconfig()
```

```python
%%ipytest



# Write your code here
def add_spaces(input_string):
    # Initialize an empty string to hold the result
    result = ""
    
    # Loop over the index of each character in the input string
    for i in range(len(input_string)):
        result = result + input_string[i]  # Add the current character to the result
        if i < len(input_string) - 1:  # Check if it's not the last character
            result += " "  # Add a space after the character
    
    return result  # Return the final string with spaces


def test_empty_string():
    result = add_spaces("")
    assert result == "", "Expected an empty string to return an empty string, but got " + str(result)

def test_single_character():
    result = add_spaces("A")
    assert result == "A", "Expected 'A' to return 'A', but got " + str(result)

def test_two_characters():
    result = add_spaces("AB")
    assert result == "A B", "Expected 'AB' to return 'A B', but got " + str(result)

def test_multiple_characters():
    result = add_spaces("Python")
    assert result == "P y t h o n", "Expected 'Python' to return 'P y t h o n', but got " + str(result)
```

## Calculating Primes

This requires a few steps. The key steps are as follows:
* We can build a list of primes by starting with an empty list then calling the `append()` method to add each prime number we find to the list. This means we need to start with an empty list before any loops.
* We can use a `for` loop to iterate over the range of numbers from 2 to the input number. As we want the range to include the input number, we need to use `n + 1` as the upper limit of the range.
* For each number in the range we can initially assume it is prime, then use another `for` loop to iterate over all the primes we've found so far, stored in `result`. If the `candidate` is divisible by any of these primes, it is not prime, so we can set `is_prime` to `False`.
* After the inner loop finishes, we can check the value of `is_prime`. If it is still `True`, we can add the `candidate` to the list of primes using the `append()` method.
* After the outer loop finishes, we can return the list of primes.

```python
import ipytest
ipytest.autoconfig()
```

```python
%%ipytest

# Write your code here
def calculate_primes(n):
    result = []
    for candidate in range(2, n + 1):
        is_prime = True
        for i in result:
            if candidate % i == 0:
                is_prime = False
                break
        if is_prime:
            result.append(candidate)
    return result


def test_no_candidates():
    result = calculate_primes(0)
    assert result == [], "There are no prime numbers less than or equal to 0 so an empty list should have been returned, but the value returned was " + str(result)

def test_one():
    result = calculate_primes(1)
    assert result == [], "There are no prime numbers less than or equal to 1 so an empty list should have been returned, but the value returned was " + str(result)

def test_two():
    result = calculate_primes(2)
    assert result == [2], "The only prime number less than or equal to 2 is 2, so a list containing only 2 should have been returned, but the value returned was " + str(result)

def test_three():
    result = calculate_primes(3)
    assert result == [2, 3], "The prime numbers less than or equal to 3 are 2 and 3, so a list containing these two numbers should have been returned, but the value returned was " + str(result)

def test_twenty():
    result = calculate_primes(20)
    assert result == [2, 3, 5, 7, 11, 13, 17, 19], "The prime numbers less than or equal to 20 are 2, 3, 5, 7, 11, 13, 17 and 19, so a list containing these numbers should have been returned, but the value returned was " + str(result)
```
