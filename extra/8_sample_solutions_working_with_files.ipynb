# Working with files


## File objects

When we want to read or write data to a file, we use the function `open()` to create a file object that we can work with.


```python
f = open('data/example.txt', 'r')
```

Notice that we provide the path to the file that we want (as a string containing either a relative or absolute path). The second argument `'r'` shows that we want to read from this file. Have a look at the contents of this file.

Having opened the file, we can read data from it. Usually the most useful way to read text data is line-by-line. The `readline()` method will collect data from the file up to and including the next newline (`'\n'`) character, update the "file pointer" to the new position and return the data read as a string. The next time that we call `readline()`, we will get the following line of text.

```python
first_line = f.readline()
second_line = f.readline()
print(first_line)
print(second_line)
```

Notice that the newline character is included in the string that is returned.

When we are finished with a file, we need to close it to free up the system resources.

```python
f.close()
```

Now we can't read anything else from the file until we open it again.

```python
f.readline()
```

## `with`

To make sure that the file is always closed correctly, even when an exception occurs, it is good practice to use a special python structure called a `with` block when working with files.

```python
with open('data/example.txt', 'r') as f:
    first_line = f.readline()
    second_line = f.readline()
    print(first_line)
    print(second_line)
```

With this structure, the file is automatically closed when the program leaves the block. You can check that f is now closed:

```python
f.closed
```

## Iterating with files

A convenient way of working with text files is to iterate over the file object directly. This is equivalent to reading line by line until there are no more lines in the file.

```python
with open('data/example.txt', 'r') as f:
    for line in f:
        print(line)
```

To make an output file, we swap `'r'` for `'w'` when opening the file, and use the `write()` method to send text to the open file.

```python
with open('output/my_output_file.txt', 'w') as f:
    f.write('He who laughs at himself\n')
    f.write('never runs out of things to laugh at.\n')
```

After running the code above, check that the file was created correctly.

Be careful with opening files for output, as you will destroy any existing file with the same name.

## Worked example: FASTA format

[FASTA format](https://en.wikipedia.org/wiki/FASTA_format) is a simple text format for storing biological sequences such as DNA, RNA and proteins.

A record in a FASTA file consists of a single description line starting with the `>` character, followed by the sequence itself (single letters indicating amino acids (for protein) or nucleic acids (for DNA/RNA)). The sequence can be split over several lines of text, which can be any length.


The file `data/example.fasta` contains a single protein sequence in FASTA format.

### 1.
Write a function that accepts the path to a file containing a single FASTA-format sequence, reads it and returns the data as a dictionary where the key is the contents of the sequence description line (minus the `'>'`) and the value is the sequence itself (without spaces or newlines).

```python
def read_one_fasta(path):
    with open(path, 'r') as f:
        descr = f.readline()[1:].rstrip()
        seq = ''
        for line in f:
            seq = seq + line.rstrip()
        return {descr: seq}
```

```python
read_one_fasta('../course_content/data/example.fasta')
```


A FASTA file can contain multiple records.


### 2.
Write a function that accepts the path to a file containing multiple FASTA-format sequences, reads it and returns the data as a dictionary where the keys are the contents of the sequence description lines (minus the `'>'`) and the values are the sequences (without spaces or newlines).

```python
def read_multi_fasta(path):
    fasta_dict = {}
    with open(path, 'r') as f:
        descr = f.readline()[1:].rstrip()
        seq = ''
        for line in f:
            if(line.startswith('>')):
                fasta_dict[descr] = seq
                descr = line[1:].rstrip()
                seq = ''
            else:
                seq = seq + line.rstrip()
    return fasta_dict
```

```python
read_multi_fasta('../course_content/data/multi_example.fasta')
```

## 3.

The file `data/S_cerevisiae_ORFs.fasta` contains the open reading frames (ORFs, DNA sequences encoding proteins) from the yeast *Saccharomyces cerevisiae*.

Using your `base_counter()` function from the Dictionaries notebook (or from the solutions folder), make a dictionary storing the GC content (percentage of the sequence that is guanine `G` or cytosine `C`) for each sequence. GC content is related to the thermal stability of a DNA molecule.


```python
# copied from sample_solutions_dictionaries.ipynb

def base_counter(sequence):
    base_counts = {"A": 0, "C": 0, "G": 0, "T": 0}
    for base in sequence:
        base_counts[base] += 1
    return base_counts
```

```python
db = read_multi_fasta('../course_content/data/S_cerevisiae_ORFs.fasta')
```

```python
gc_content = {}
for descr in db:
    counts = base_counter(db[descr])  # count the bases for this sequence
    gc_content[descr] = (counts['G'] + counts['G']) / len(db[descr])

```

## 4.
Create an output file where the first column is the gene symbol (the first identifiers in the ORF data file, starting `YAL`) and the second column is the GC content of that sequence. Separate the columns with a comma.

```python
with open('../course_content/output/yeast_GC.txt', 'w') as f_out:
    for descr in gc_content:
        gene = descr.split(' ')[0]
        f_out.write(gene + ',' + str(gc_content[descr]) + '\n')
```

## 5.

What is the average GC content for an *S. cerevisiae* ORF?

```python
tot = 0
for gc in gc_content.values():
    tot = tot + gc

print(tot / len(gc_content))
```
