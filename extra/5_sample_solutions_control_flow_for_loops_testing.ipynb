## Count Numbers Divisible by 7

There are several things we need to get right to get the correct answer:

* We must initialise the accumulator variable (`x` in the sample solution below) to 0 before the loop starts.
* We must use the correct range of numbers. The `range` function is inclusive of the first number and exclusive of the second number, so we need to use `range(500, 701)` to include 700 in our loop.
* We need a nested `if` statement to check if the current number is divisible by 7. We can do this using the modulus operator `%`. If `i % 7 == 0`, then `i` is divisible by 7.
* We need to increment the accumulator variable by 1 each time we find a number that is divisible by 7.
* The final `print` statement should be outside of the loop, so that it only executes once after the loop has finished.

```python
x = 0

for i in range(500, 701):
    if i % 7 == 0:
        x += 1

print(x)
```

## Using Tests to Debug Code

There were two problems with the original code:

1. The `range` function was not set up to return the value of `end` (remember `range`s are exclusive of the second value).
2. The code was summing the squares of even numbers. This meant it worked fine when the value of `start` was even, but not when it was odd. The easiest solution was to use the third argument of the `range` function to set a step of 2, which will return every other number beginning with the first number. This will work regardless of whether `start` is even or odd.

```python
import ipytest
ipytest.autoconfig()
```

```python
%%ipytest



def sum_of_squares_every_other(start, end):
    total = 0
    for i in range(start, end + 1, 2):
        total += i ** 2
    return total

def test_no_values():
    result = sum_of_squares_every_other(2, 1)
    assert result == 0, "If the first number is larger than the second, the sum of every other square should be 0 but it was "+ str(result)

def test_zero_to_zero():
    result = sum_of_squares_every_other(0, 0)
    assert result == 0, "The sum of every other square from 0 to 0 should be 0 but it was "+ str(result)

def test_zero_two():
    result = sum_of_squares_every_other(0, 2)
    assert result == 4, "The sum of every other square from 0 to 2 should be 4 (0^2 + 2^2) but it was "+ str(result)

def test_two_to_ten():
    result = sum_of_squares_every_other(2, 10)
    assert result == 220, "The sum of every other square from 2 to 10 should be 220 (2^2 + 4^2 + 6^2 + 8^2 + 10^2) but it was "+ str(result)

def test_one_to_five():
    result = sum_of_squares_every_other(1, 5)
    assert result == 35, "The sum of every other square from 1 to 5 should be 35 (1^2 + 3^2 + 5^2) but it was "+ str(result)
```

# Writing Tests

There are lots of different tests that could be written for this function. Here are a few examples:

```python
import ipytest
ipytest.autoconfig()
```

```python
%%ipytest

def count_common_multiples(start, end, factor1, factor2):
    count = 0
    for i in range(start, end + 1):
        if i % factor1 == 0 and i % factor2 == 0:
            count += 1
    return count

# Write your tests here
def test_start_higher_than_end():
    result = count_common_multiples(10, 1, 2, 3)
    assert result == 0, "If the start is higher than the end, the count should be 0"

def test_no_common_multiples():
    result = count_common_multiples(1, 10, 2, 7)
    assert result == 0, "There should be no common multiples of 2 and 7 between 1 and 10"

def test_one_common_multiple():
    result = count_common_multiples(1, 10, 2, 3)
    assert result == 1, "There should be one common multiple of 2 and 3 between 1 and 10 (which is 6)"

def test_factors_the_same():
    result = count_common_multiples(1, 10, 2, 2)
    assert result == 5, "There should be five common multiples of 2 and 2 between 1 and 10 (which are 2, 4, 6, 8, 10)"

def test_multiple_multiples():
    result = count_common_multiples(1, 30, 3, 5)
    assert result == 2, "There should be two common multiples of 3 and 5 between 1 and 30 (which are 15 and 30)"

def test_large_range():
    result = count_common_multiples(1, 100, 4, 6)
    assert result == 8, "There should be eight common multiples of 4 and 6 between 1 and 100 (which are 12, 24, 36, 48, 60, 72, 84, 96)"
```
