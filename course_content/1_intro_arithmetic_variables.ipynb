### More operators

Alongside the above four, there are three more you may find use for.

| Operator | Effect |
|------|-------|
| `+` | Adds two numbers together |
| `-` | Subtracts the number on the right from the number on the left |
| `*` | Multiplies two numbers together |
| `/` | Divides the number on the left by the number on the right |
| `//` | Divides, then rounds the result down to the nearest whole number |
| `%` | Returns the remainder after dividing the number on the left by the number on the right |
| `**` | Raises the number on the left to the power of the number on the right |

Run the cell below to see all seven applied to the same pair of numbers.

`````python
print(10 + 3)
print(10 - 3)
print(10 * 3)
print(10 / 3)
print(10 // 3)
print(10 % 3)
print(10 ** 3)
`````

Let's take a closer look at two of these.

`//` is floor division. `10 / 3` is `3.333...`, so `10 // 3` is `3`. It discards everything after the decimal point rather than rounding to the nearest whole number.

`%` is the modulo operator, and gives the remainder. `3` goes into `10` three times with `1` left over, so `10 % 3` is `1`. This is more useful than it first looks: a number is even if `number % 2` is `0`, and divisible by 7 if `number % 7` is `0`. You will use that idea later in the course.

One more thing to notice. Run the cell below.

`````python
print(10 / 2)
print(10 // 2)
`````

`10 / 2` gives `5.0`, while `10 // 2` gives `5`.

Python treats whole numbers and decimal numbers as two different kinds of data, and ordinary division always gives a decimal result, even when the answer is a whole number. We will look at this properly in the next session.

### Order of operations

Python follows the usual operator precedence that applies in mathematics.

| Priority | Operators |
|------|-------|
| 1 | Parentheses `()` |
| 2 | Exponentiation `**` |
| 3 | `*`, `/`, `//`, `%` |
| 4 | `+`, `-` |

As expected, you can use parentheses (round brackets) to force a specific order of evaluation.

Run the cell below and check that each result is what you would predict.

`````python
print(2 + 3 * 4)      # The multiplication happens first
print((2 + 3) * 4)    # The parentheses force the addition to happen first

print(2 * 3 ** 2)     # The exponentiation happens first
print((2 * 3) ** 2)   # The parentheses force the multiplication to happen first
`````

Even when parentheses aren't strictly needed, add them where they help with the *readability* of the code. `(a * b) + c` and `a * b + c` do exactly the same thing, but the first makes it explicit for whoever is reading it later.

> On whitespace: Python ignores the spaces around operators, but people do not. The convention is a single space either side of an operator, and no space directly inside parentheses. `(2 + 3) * 4` is easier to read than `(2+3)*4`.

## Variables

Doing a calculation is only useful if you can hold on to the result. A variable is a name that refers to a piece of data, so that you can use it again later.

To assign data to a variable in Python, use a single equals sign `=`. The name goes on the left and the value goes on the right.

Run the cell below.

`````python
sample_count = 6
total_volume_ml = 12 * 2
volume_per_sample_ml = total_volume_ml / sample_count
`````

Notice that assigning data to a variable produces no output at all. The cell ran, but there was nothing to display.

To see the value of a variable, print it.

`````python
print(sample_count)
print(total_volume_ml)
print(volume_per_sample_ml)
`````

A variable can be assigned the value of another variable. The value is read out of the one on the right and assigned to the one on the left, leaving the original untouched.

`````python
measured_length_cm = 12.4
recorded_length_cm = measured_length_cm

print(measured_length_cm)
print(recorded_length_cm)
`````

### Assignment is not equality

The `=` sign does not behave like an equals sign in an equation. It means: *work out the value on the right-hand side, then attach the name on the left-hand side to the result*.

The following is invalid in Python, even though it is a perfectly good algebraic statement.

```python
3 = x
```

The following, though, **is** valid Python, despite being true only for $x = 0$ algebraically:

```python
x = x + x
```

The right-hand side is worked out first, using the current value of `x`. That result then becomes the new value of `x`.

`````python
x = 1
x = x + x
print(x)
`````

Counting things up this way is extremely common, so Python provides a shorthand. `count += 1` means exactly the same as `count = count + 1`. There are matching shorthands for the other operators: `-=`, `*=` and `/=`.

`````python
reading_count = 0

reading_count += 1
reading_count += 1
reading_count += 5

print(reading_count)
`````

### Reassigning variables

You can assign to the same variable as many times as you like, but each new assignment overwrites the previous value. Once a value has been overwritten, it is gone.

Notice how variables are happy to hold different types of data. A variable holding a number can later hold text instead. This isn't true in every programming language.

`````python
sample_label = 12.5
print(sample_label)

# Assign a new value to the same name - the 12.5 is now lost
sample_label = "Site B, replicate 3"
print(sample_label)
`````

## Naming variables

There are a few rules about variable names in Python:

* Names must start with a letter or an underscore `_`
* Names can only contain letters, numbers and underscores
* Names are case sensitive, so `sample_id` and `Sample_ID` are two different variables
* Names may not be one of Python's reserved keywords

The reserved keywords are:

| | | | | |
|-|-|-|-|-|
| `False` | `None` | `True` | `and` | `as` |
| `assert` | `async` | `await` | `break` | `class` |
| `continue` | `def` | `del` | `elif` | `else` |
| `except` | `finally` | `for` | `from` | `global` |
| `if` | `import` | `in` | `is` | `lambda` |
| `nonlocal` | `not` | `or` | `pass` | `raise` |
| `return` | `try` | `while` | `with` | `yield` |

You do not need to memorise this list. VS Code colours keywords differently from ordinary names, so you will notice when you have used one by accident. Run the two cells below to see what happens when a name breaks the rules.

### Choosing good names

The rules tell you what Python will accept. They say nothing about what makes a name *good*, and that is the part that matters.

Names like `x`, `y` and `data` are legal but tell a reader nothing. A good name says what the value is, and where it helps, what units it is in.

Consider the difference:

* `t` versus `elapsed_time_s`
* `TempData` versus `soil_temperature_c`
* `conc` versus `glucose_concentration_mmol_l`
* `n` versus `participant_count`

The convention in Python is to write variable names in lower case, separating words with underscores. This is called snake case.

You are writing for two audiences: whoever reads your code next, and yourself in six months. Neither of them will remember what `x2` was for.

## When Things Go Wrong

You will produce errors constantly, and that is completely normal. What matters is being able to read them.

When Python cannot run your code, it prints a block of text called a traceback. **The last line is the one that tells you what went wrong.** Everything above it describes where the problem happened.

You have already met two kinds:

* `SyntaxError` means the code is not valid Python at all, so nothing in the cell runs.
* `NameError` means Python met a name it does not recognise.

Here is a third.

`````python
print(10 / 0)
`````

`ZeroDivisionError: division by zero` says exactly what happened.

When you hit an error you don't understand, read the last line, then look at the line the traceback points to. If that doesn't help, search for the text of the error message.

## Appendix: Text Cells and Markdown

This section is here for reference. You do not need to work through it today.

The text cells in this Notebook are written in a formatting language called Markdown. Double-click on this cell. The formatting disappears and you see the raw text, including the symbols that produce the formatting. Press `Shift` and `Enter` to render it again.

The formatting comes from a language called Markdown. Here is most of what you will need.

### Headings

Headings are made with hash `#` symbols. Do not confuse these with comments in Python - those only apply inside code cells.

```
# Big Heading
## Smaller Heading
### Smaller Still
```

There are six levels, from one hash down to six.

### Emphasis and code

```
*italic text* uses single asterisks
**bold text** uses double asterisks
`inline code` uses single backticks
```

Larger pieces of code go in a fenced block, opened and closed with three backticks. Naming the language after the opening backticks turns on syntax colouring:

````
```python
sample_count = 6
```
````

### Lists

Lists are made with asterisks or dashes. Nested items are indented, and there should be a blank line before the list starts.

```
* first item
    * nested item
* second item
```

### Tables

Columns are separated by pipes `|`, and the second row marks the header off from the body. The columns do not need to line up in the raw text.

```
| Column 1 | Column 2 |
| - | - |
| A | B |
| C | D |
```

Which renders as:

| Column 1 | Column 2 |
| - | - |
| A | B |
| C | D |

### Mathematics

Mathematical notation is written in LaTeX, between dollar signs.

```
$\frac{\alpha}{2} = 7$
```

Which renders as $\frac{\alpha}{2} = 7$.

This lets you state the equation a piece of code implements, right next to the code itself.
