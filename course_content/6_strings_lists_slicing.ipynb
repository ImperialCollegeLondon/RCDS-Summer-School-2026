# Strings, Lists and Slicing



## Strings

A string is a collection of characters, such as letters, numerals, punctuation marks, and whitespace. Strings are useful for holding text and qualitative data, displaying messages to users, and for storing data that is read from files. Strings are immutable, which means that once a string is created, it cannot be changed. We can create a string by enclosing characters in either single quotes (`'`) or double quotes (`"`). For example:

```python
print("Hello, World!")

x = 'Python is fun!'
print(x)
```

### Indexing

Each character in a string has an index, which is its position in the string. The first character **has an index of 0**, the second character has an index of 1, and so on. We can access individual characters in a string by writing the name of the variable that holds the string, followed by the index of the character we want to access in square brackets. For example:

```python
my_string = "Hello, World!"

print(my_string[0])
print(my_string[1])
print(my_string[5])
```

We can also use negative indexing to access characters from the end of the string. The last character has an index of -1, the second-to-last character has an index of -2, and so on. For example:

```python
my_string = "Hello, World!"
print(my_string[-1])
print(my_string[-2])
print(my_string[-5])
```

### Concatenation

Strings can be concatenated (joined together) using the `+` operator. This produces a new string and leaves the original strings unchanged. For example:

```python
a = "Hello "
b = "there!"

c = a + b
print(c)

# The original strings are unchanged
print(a)
print(b)
```

We can also multiply strings using the `*` operator. This produces a new string that is the original string repeated a specified number of times. For example:

```python
x = "Badger "
y = x * 5
print(y)
```

### f-strings

When creating or printing a string, you will sometimes want to include the value of a variable in the string. One way to do this is to use an f-string, which is a string that starts with the letter `f` and contains placeholders for variables inside curly braces `{}`. For example:

```python
a = 1
b = 2
x = f"a: {a}, b: {b}"
print(x)

my_name = "Alice"
print(f"Hello, {my_name}!")
```

### New Lines

When creating a string, you can add the characters `\n` to indicate a new line. For example:

```python
x = "Hello\nWorld"
print(x)
```

### Length of a String

We can also calculate the number of characters in a string using the built-in `len()` function. For example:

```python
x = "A B!"
print(len(x)) # Note whitespace and punctuation count
```

### Testing Membership

We can test whether a string contains a certain character or substring using the `in` operator. This returns a boolean value (`True` or `False`). For example:

```python
print("fun" in "fundamentals")
print("fun" in "skipping ahead")

# The in operator can be used in an if statement to test membership and execute code conditionally
x = "strings are sequences"
if "sequences" in x:
    print("Yes, 'sequences' is in the string x")
```

## Lists

Lists are a type of data structure in Python that can hold an ordered collection of items. They can hold any number of items of any type. We can create a list by placing a comma-separated sequence of items inside square brackets `[]`. For example:

```python
x = [1, 2, 3]
print(x)

# An empty list
y = []
print(y)

# Creating a list using variables
z = [x, 2.3, y, "Hello"]
print(z)
```

In the list above, we created lists that contained mixtures of integers, floats, strings and even other lists. We can do many of the same operation on lists that we could do with strings, such as indexing, concatenation and multiplication. For example:

```python
my_list = [10, 20, 30]

# Indexing
print(my_list[0])
print(my_list[-1])

# Concatenation
x = my_list + [40, 50]
print(x)

# Multiplication
y = my_list * 3
print(y)

# Finding the length
print(len(my_list))

# Checking membership with in
print(3 in [1, 2, 3, 4, 5])
```

### Modifying Lists

We can modify lists through a number of methods. The simplest of these is item assignment, which allows us to change the value of an item in a list by accessing it through its index. For example:

```python
my_list = [1, 2, 3, 4, 5]
my_list[0] = 10
print(my_list)
```

In the example above, we wrote the name of the variable referencing the list, followed by the index of the item we wanted to change in square brackets, then the assignment operator `=` and the new value we wanted to assign to that position in the list. This changed the value of the first item in the list from `1` to `10`.

We can also add items to a list using the `append()` method of the list. To do this, we write the name of the variable referencing the list, followed by a dot `.` and the name of the method, then parentheses containing the item we want to add to the list. This will modify the initial list. For example:

```python
my_list = [1, 2, 3, 4, 5]
my_list.append(6)
print(my_list)
```

We can also insert a new item at a specific index in the list using the `insert()` method. This method takes two arguments: the index at which to insert the new item and the item itself. This modifies the initial list and shifts all entries after the index where the new value is being inserted one to the right. For example:

```python
my_list = [1, 2, 3, 4, 5]
my_list.insert(2, 99)
print(my_list)
```

## Looping Over Lists and Strings

In Python, we can use a `for` loop to iterate over more than just a `range` of numbers. We can iterate over any object which is "iterable", including lists and strings. When we do this, the loop variable takes on the value of each item in the list or string in turn. For example:

```python
# Loop over a string
for char in "Hello":
    print(char)

# Loop over a list
my_list = [1, 2, 3, 4, 5]
for item in my_list:
    print(item)
```

## Mutability

One key difference between strings and lists is that strings are immutable, while lists are mutable. This means that once a string is created, it cannot be changed, while a list can be modified after it is created. This is why it's fine to use item assignment on a list, but a similar operation doesn't exist for strings. For example, the following code will raise an error:

```python
my_list = [1, 2, 3, 4, 5]
my_list[0] = 10 # This is valid because lists are mutable
print(my_list)

my_string = "Python"
my_string[0] = "J" # This will raise an error because strings are immutable
```

This gives rise to some interesting behaviours with mutable objects when multiple variables reference the same object. For example, consider the following code:

```python
x = [1, 2, 3]
y = x # This means that y is now a reference to the same list object as x

x[0] = 10 # This modifies the list that both x and y reference
print(x)
print(y)
```

This means that, when we modify the contents of a mutable object, all variables that reference that object will see the change. This is not the case for immutable objects, which cannot be changed after they are created.

Note that, for both mutable and immutable objects, if we assign a new value to a variable, it will no longer reference the original object, but a new one instead. For example:

```python
x = [1, 2, 3]
y = x # y is a reference to the same list object as x
x = [4, 5, 6] # This reassigns x to a new list object
print(x)
print(y)

# Same for strings
x = "Hello"
y = x # y is a reference to the same string object as x
x = "Goodbye" # This reassigns x to a new string object
print(x)
print(y)
```

The important things to remember are:
* Assigning a new value to a variable does not modify the original object; it simply makes the variable reference a new object.
* Mutable objects, on the other hand, can be modified in place, and all references to the object will see the change.
* Immutable objects cannot be modified in place, and any attempt to do so will result in an error.
