There are a number of variable types in Python - numerical types, text (or strings), Boolean (True, False) and collections of variables (those will be covered later). It's also possible to define your own variable types. Every variable in Python references an instance of a particular type of data.

You can check which type of object any variable references using the ```type``` command. Python responds by showing a ```class``` that is used to define the object. Below, you can see objects of class ```integer``` (whole number), ```float``` (decimal number), ```bool``` (Boolean value), and ```str``` (string or text).

````python
print(type(1))
print(type(1.1)) 
print(type(True))
print(type("Hello World"))
````

Inside Python, everything is treated as an "objects". Objects of different types will have different properties and behave differently. Predefined data types also come with allowed operations that can be done on them.

## 1. Numeric values
A family of types (```int```, ```float``` and ```complex```) deals with different numeric types - integers, decimal numbers and complex numbers.

Below are three variables that are integers. You don't have to do anything special to inform Python about the type, it will decide on the type for you. Integer objects have various operations associated with them, for example subtraction. If you subtract two integers, the resulting value is also an integer.

````python
int1 = 4
int2 = 3
int3 = 2

print(type(int1))
print(type(int2))
print(type(int3))

print("Subtraction:")
print(int1 - int2)
print(type(int1 - int2))
print(int3 - int2)
print(type(int3 - int2))


````

However, operations using one type of number will not always produce the same type of number. For instance, division of two integers will produce a ```float``` type, even when the result is expressible as an integer (for instance, $4/2=2$).

This may sound complicated, but the good news is that Python generally does smart things when working with the numerical types so you don't need to worry about the conversions most of the time.

For floats, Python will automatically use different precisions and number of decimal places, allowing a very wide range of values to be represented. 

````python
print("Division:")
int1 = 4
int2 = 3
int3 = 2

print(int1 / int2)
print(type(int1 / int2))

print(int1 / int3)
print(type(int1 / int3))
````

You can also perform "floor division". Unlike regular division, floor division rounds down to the nearest integer. This behavior applies to both positive and negative numbers - for example, ```5 // 2``` results in ```2```, while ```-5 // 2``` results in ```-3```. If you divide two integers using floor division, the result will always be an integer. If at least one number is a float, the result will be a float.

````python
print("Floor Divisions:")
int1 = 4
int2 = 3
int3 = 2

print(int1 // int2)
print(type(int1 // int2))

print(int1 // int3)
print(type(int1 // int3))

print(1 // 2)
print(type(1 // 2))

#notice how floor division rounds down to the nearest integer for negative numbers
print(-1 // 2)
print(type(-1 // 2))

#notice how floor division works with floats as well, however the result is of type float
print(1.0 // 2)
print(type(1.0 // 2))
````


### Type conversions

Sometimes, you will want to convert between different ```int```s and ```float```s. You may obtain a float version of the variable ```x``` by writing:

```python
float(x)
```

By writing:

```python
int(x)
```

you will convert a number to an integer (rounding toward zero). 

````python
print(int(3.6))
print(int(-2.9))
print(float(2))
````


### Rounding numbers

If you want to round to the nearest number, you can use the ```round``` function:

```python
round(x)
```

Notice that rounding of numbers exactly halfway between two integers follows the "round half to even" strategy.

```python
print(round(2.5) == 2)
print(round(3.5) == 4)
```


````python
print(round(3.6))
print(round(-3.6))
print(round(2.5))
print(round(3.5))   
````

### Interrogating strings

The string object comes with many ways of interrogating its value. These are called methods that are associated with the string class. 

You will notice that we use methods by appending the method name to the variable using a dot, for example: ```a.count("e")```, which will count occurrences of letter e in the string that is stored in variable a. ```count``` is a method, ```("e")``` is an argument to the method. If there is no argument, a method is followed by empty parentheses ```()```.

Common string methods fall into categories:

- Case transformation by capitalisation or lower case
- Searching such as counting occurrences or finding a substring within a string
- Validation - testing content for numbers, letters, upper or lower case
- Text alteration, trimming, padding - for example, removing leading and trailing whitespace 
- Joining and splitting - creating one string from several shorter ones, splitting by a specified delimiter

Below is an example. You should always refer to documentation for a complete list. Last time I checked, there were 47 methods associated with the string object.

Counting occurrences - the ```count``` method returns how many times one string appears in another

````python
a = "Strings are great!"
print(a.count("e"))
print(a.count("ea"))
print(a.count("s")) # Note matches are case-sensitive
print(a.count("!")) # Can count numerals or punctuation characters
````

### Combining string functions

Often, you may wish to combine several string methods to achieve a desired outcome. For example, you may want to turn a list of space-delimited words into lowercase, strip any trailing spaces and add a comma as a delimiter. You can do each operation separately and assign the result to a new variable. This is possible but unnecessary. 

````python

s = "Rabbits RAbbits RABBITS  "
a = s.lower()
b = a.rstrip()
c = b.replace(' ',',')
print(c)
````

Instead, you can use the following compact way.

````python

s = "Rabbits RAbbits RABBITS  "
print(s.lower().rstrip().replace(' ',','))
````

## Mixing data types

What happens if you inadvertently try to perform an operation on two different data types? It depends: Python may adapt and choose the best type for the output (if that makes sense), do something unexpected, or produce a type error.

Let's explore some examples.


If you add ```int``` and ```float```, Python stores the result in ```float```.

````python
a = 5 + 1.1
print(a)
print(type(a))
````

If you try to add ```int``` and string ```str```, Python does not try to guess what it is that you may be trying to do. It produces a ```TypeError```. 

````python
str1 = "Hello"
int1 = 5
print(str1 + int1) # Convert int to str before concatenation
````

Note: The error suggests that you can use addition on strings. That is true, you can "add" strings and it will result in concatenation.

````python
a = "Hello"
b = "world!"
print (a + " " + b) # Concatenate strings with a space in between
````

Similarly, you can multiply strings. This will result in concatenation.

````python
print(3 * "Bye")
````

The last example deals with an "unexpected" outcome. If you add an ```int``` and ```bool```, Python considers True to be 1 and False to be 0 and will happily use them as integers.

````python
print(10 + True)
print(10 + False)
````

### Using numerical types in string concatenation

If you'd like to use concatenation of a few variables, they all have to be of type ```str```. Use type conversion with the ```str``` function to convert numerical types to strings. 

````python
speed = 60.0
units = "km per hour"

print("The speed is " + str(speed) + " " + units + ".") # Convert float to str before concatenation
````
