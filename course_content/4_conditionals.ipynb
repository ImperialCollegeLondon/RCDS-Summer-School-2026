# Conditionals

In this section, we will look at ways of controlling the flow of instructions through your program using conditionals. Generally, conditionals (and loops that you will see later) are called control structures. Without control structures, the program would execute in sequence from the first to the nth line. Conditionals let you make a decision to execute a specific piece of code only if some condition is met.

Conditionals involve an expression that results in either ```True``` or ```False``` (that is of Boolean type). For example:

```python
a > 0
  ```

In a conditional, a boolean expression is preceded by the keyword ```if``` and followed by ```:```. The expression evaluates to ```True``` or ```False```. This, in turn, determines which block of code is executed. The body of the code block is indented.

In the following code, the print statement is executed only if ```a > 0``` evaluates to ```True```.


```python
if a > 0:
  print("a is positive")
```

In the code cell below, run the code. Then, introduce indentation error and rerun. What kind of error did you get?

````python
a = 2

if a > 0:
    print("a is positive")
````

### Else block
So far, we were able to execute code only if the boolean expression was evaluated to ```True```. In the following code, another block of code is added that starts with ```else```. This is executed if ```a > 0``` is evaluated to ```False```.

```python
if a > 0:
  print("a is positive")
else:
  print("a is either zero or negative")
```
Below is a code example, examine it and set the ```number``` variable to a few different values.

````python
number = -7

# Check if the number is greater than or equal to zero
if number >= 0:
    print("The number is positive.")
else:
    print("The number is negative.")
````

## Boolean Operators

Before we talk more about ```if``` statements, it will be useful to know how to compare different variables to evaluate the expression following the ```if``` keyword.

To do this, we can use conditional operators, each of which returns a ```bool``` value of either ```True``` or ```False```. These include:

| Operator | Example  | Description |
|------|-------|-------|
| ```<``` | ```a < b``` | Is true if the value of ```a``` is less than the value of ```b``` |
| ```>``` | ```a > b``` | Is true if the value of ```a``` is greater than the value of ```b``` |
| ```==``` | ```a == b``` | Is true if the values of ```a``` and ```b``` are equal |
| ```!=``` | ```a != b``` | Is true if the values of ```a``` and ```b``` are not equal |
| ```in``` | ```a in b``` |  If ```b``` is a collection (such as a list), checks if any entry of ```b``` has the same value as ```a``` (collections will be covered later) |

Here are some examples:

````python
print("Less than -----------------")
print(1 < 2)
print(1 < 1)
print(2 < 1)
````

````python
print("Greater than---------------")
a = 1
b = 2
print(a > 2)
print(1 > a)
print(b > a)
````

````python
print("Equality-------------------")
a = 1
b = 2
print(a == 1)
print(a == b)
````

````python
print("Inequality------------------")
a = 1
b = 2
print(a != 1)
print(a != b)
````

### Single or double equal sign?

Assigning a value to a variable is done with a single ```=```, while checking equality requires a double ```==```. Using a single equal sign in a boolean expression causes a syntax error.

## Combining boolean expression using logical operators

We can also combine booleans using operators. A selection of boolean operators is explained below.

| Operator | Example  | Description |
|------|-------|-------|
| ```not``` | ```not a``` | Is true if the value of ```a``` is ```False``` |
| ```and``` | ```a and b``` | Is true if both ```a``` and ```b``` are ```True```|
| ```or``` | ```a or b``` | Is true if either the value of ```a``` or ```b``` is ```True``` |


We can also combine booleans in a particular order using parentheses, which work the same as for numeric operations. The order of precedence for these operations is:

* Parentheses
* Comparison (e.g. ```>```)
* ```not```
* ```and```
* ```or```


Here is an example of a combined boolean expression:
```python
if a > 0 or b > 0:
  print("one of the values is positive")
```

## Truthiness and falsiness

A single value in a boolean expression can also be evaluated. Values like 0, empty strings, empty lists, and None behave as False in conditions, while most other values behave as True.

Below are a few examples:

````python
if 0:
    print("True. This will not be printed because 0 is considered False in Python.")
else:
    print("False. This will be printed because 0 is considered False in Python.")

if "A":
    print("True. This will be printed because non-empty strings are considered True in Python.")

if None:
    print("True. This will not be printed because None is considered False in Python.")
````

## If Block Syntax

An ```if``` block allows us to use these comparisons to decide which code to execute. It has the syntax:

```python
if [boolean expression]:
  [Do some stuff spanning...
  ...multiple lines]
elif [boolean expression]:
  [Do different stuff]
elif [boolean expression]:
  [Do other different stuff]
else:
  [Do other other different stuff]

[Code to be executed afterward]
  ```

Python will work through the conditional block from top to bottom, checking whether each ```bool``` expression is ```True``` or not. If one is ```True```, it executes the corresponding indented block and then skips to the bottom of the ```if``` block (here, this is after the indented block following the ```else``` statement). If none of the boolean expressions are ```True```, it executes whatever is in the else statement (if one is present).

You can include as many ```elif``` ("else if") statements as you like, but you aren't required to include any. You are also not required to include an ```else``` statement but, if you do, you can include only one. This means at most one set of code from an if block will be executed. If an else statement is present, then exactly one set will be executed.
