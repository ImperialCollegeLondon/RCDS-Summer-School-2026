# Dictionaries and Basic IO

A dictionary is a data structure that maps hashable keys — such as strings, integers, and floats - to associated values. Values in dictionaries can be any type. Dictionaries can be very useful for storing data, which is best identified by something other than a series of consecutive integers as with a list.

Input/Output (I/O) is how programme interacts with the outside world: receiving information from a user or file (Input) and displaying or saving results (Output). For example, `print()` function is an often used output function.

Let's start with creating a dictionary type.

## Creating a Dictionary

To create a dictionary, we use curly braces ```{}```. If these braces are empty, the dictionary will be empty. We can also provide a series of key-value pairs, separated by commas. The key and value are separated by a colon. For example:

````python
empty_dict = {} # Create an empty dictionary
fruity_dict = {"melons": 5, "peaches": 4} # Create a dictionary with two key-value pairs

print(empty_dict)
print(fruity_dict) # When printing a dictioanry, key-value pairs are separated by commas
# Each key-value pair is represented as the key for followed by a colon and then the value
````

## Accessing Values from a Dictionary

We can access individual values of the dictionary by using the key in square brackets ```[]```. Just like values from a list, we can use these values as part of an expression. For example:

````python
print(fruity_dict["melons"]) # Access a value by using its key
b = fruity_dict["peaches"] + 5 # You can use a value in an expression
print(b)
````

If we try to access a value using a key not in a dictionary, a KeyError exception will be raised.

````python
print(fruity_dict["plums"])
````

## Updating a Dictionary

We can add new key-value pairs or update the value of existing key-value pairs by writing the name of the dictionary, followed by the key in square brackets, followed by an equals sign and the new value. For example:

````python
fruity_dict["bananas"] = 2 # Add a new key-value pair to a dictionary
fruity_dict["peaches"] = 6 # Change the value associated with an existing key
print(fruity_dict)
````

We can delete a key-value pair from a dictionary using the ```del``` keyword.

````python
del fruity_dict["melons"] # Delete a key-value pair from a dictionary
print(fruity_dict)
````

## Looping over a Dictionary

We can loop over the keys or values of a dictionary.

````python
for x in fruity_dict: # Looping over a dictionary gives the keys
    print(x)

for x in fruity_dict.values(): # To loop over the values, we need the "values" method
    print(x)
````

We can also loop over the keys and values of a dictionary at the same time using the ```items()``` method.

````python
for fruit, number in fruity_dict.items(): # To loop over both keys and values, we need the "items" method
    print(f"There are {number} {fruit}")
````

## Order of Dictionaries

Note that, for earlier versions of Python, the order of the keys may not be the same as the order in which they were added to the dictionary. This is because dictionaries are implemented using hash tables, which are not ordered. In Python 3.6 and later, dictionaries are ordered, so the order of the keys will be the same as the order in which they were added to the dictionary. However, relying on the "order" of a dictionary is not normally good practice as it can make your code less portable and tends to encourage design patterns which do not use dictionaries in the best way.

## Basic IO and Dictionaries

When combining Basic I/O with Dictionaries (dict), we can collect structured data from a user, process key-value pairs, and save or load complex records to and from files.

In Python, the `input()` function is often used to take users' input. It waits for users' input and returns the users' input as a string. 

For example:

````python
fruit = input("Enter your favorite fruit: ")
print("Your favorite fruit is:", fruit)
````

## Collecting User Input into a Dictionary

Since dictionaries store data as key-value pairs (key: value), we can use input() to prompt users for specific fields and store their responses directly under keys.


````python
# Create an empty dictionary
item_profile = {}

# Prompt user for data and assign to keys
item_profile["name"] = input("Enter item's name: ")
item_profile["stock"] = int(input("Enter item's stock quantity: "))
item_profile["price"] = float(input("Enter item's price: "))

# Output the whole dictionary
print("\nItem Profile Created:")
print(item_profile)
````

## Formatting Dictionary Output
To display dictionary data cleanly instead of printing raw brackets `{...}`, loop over the dictionary using `.items()` and format the output with f-strings.

````python
inventory = {
    "apples": 10,
    "bananas": 25,
    "oranges": 15
}

print("=== Inventory Stock ===")
# .items() gives both key (item) and value (count)
for item, count in inventory.items():
    print(f"• {item.capitalize()}: {count} in stock")
````

## Saving & Loading Dictionaries to Files (json)



Basic file IO. 

The `open()` function connects your Python code to a file on your computer. To use it, you pass two main parameters (arguments): the filename and the mode (such as "w" for write).

For example:

````python
file = open("example.txt", "w")
file.write("Hello, world!")
file.close()
````

open("example.txt", "w"): Creates or opens the file in write mode. If the file already exists, this mode overwrites its contents.

file.write(...): Sends text to the file.

file.close(): Saves the changes and releases the file back to the system.

**However**, closing files manually works, it is easy to forget. If your script crashes before reaching .close(), the file can stay locked or lose data.

To avoid this, use a context manager with the `with` keyword:

````python
with open("example.txt", "w") as file:
    file.write("Hello, world!")
````

Using `with` automatically closes the file the moment the indented block finishes—even if an error occurs. It keeps your code cleaner, safer, and easier to maintain.


Now let's work on dictionary data. Plain text file methods `file.write()` only accept strings. To save an actual Python dictionary to a file, the standard tool is Python's built-in `json` module. JSON formats dictionaries into readable text files that preserve their structure.

### Saving a dictionary

````python
import json

inventory = {
    "apples": 10,
    "bananas": 25,
    "oranges": 15
}

# Save dictionary to a .json file
with open("inventory.json", "w") as file:
    json.dump(inventory, file, indent=4)  # indent=4 makes it readable

print("Data saved successfully to inventory.json!")
````

### Loading a dictionary

````python
import json

# Read dictionary back from the file
with open("inventory.json", "r") as file:
    loaded_inventory = json.load(file)

# Now it's a real Python dictionary again!
print(type(loaded_inventory))  # Output: <class 'dict'>
print(f"Number of apples: {loaded_inventory['apples']}")  # Output: Number of apples: 10
````
