## Modules

Run the cell below. It asks Python for the square root of 16.

````python
print(sqrt(16))
````

`NameError: name 'sqrt' is not defined`. Python has no square root function built into the language.

This is deliberate. Python keeps a small core, and almost everything else lives in *modules* that you load only when you need them.

A module is a collection of Python code, normally contained in a single `.py` file. Importing a module gives you access to the functions and values defined inside it.

Python comes with a large collection of modules already installed, called the *standard library*. There are modules for maths, dates, random numbers, working with files, and more. The full list is in the [Python documentation](https://docs.python.org/3/library/).

The simplest way to import a module is to write `import` followed by its name.

````python
import math
````

Then we can use `sqrt` from inside `math`, which you reach with *dot notation*: the module name, a full stop, then the name of the thing you want.

````python
print(math.sqrt(16))
````

`math` holds a lot more than just `sqrt`. To find out what is inside a module, use `help()`. Run the cell below, then scroll through the output (there is a lot of it).

````python
help(math)
````

`help()` works on individual functions too, but is only useful once you know what you are looking for.

````python
help(math.sqrt)
````

### Using what is in a module

Access something from a module by writing the module name, a full stop, and then the name of the thing you want.

Values from a module can be used in expressions like any other variable, and functions from a module are called like any other function.

````python
# A value stored in the math module
print(math.pi)

# Using that value in an expression
angle_radians = math.pi / 2

# Calling a function from the math module
print(math.sin(angle_radians))
````

Writing `math.` in front of everything may look like a chore, but it is doing important work. It keeps the names from each module separate, so that two modules can both define something called `sqrt` without any confusion about which one you meant. This separation is called a *namespace*.

### Shortening what you type

In a file that leans on a module heavily, typing its full name in front of everything gets tedious. You can give a module a shorthand name, or *alias* as you import it.

````python
# Give the module an alias
import math as m

print(m.sqrt(2))
````

Aliases are a convenience, not a requirement. `math` has no standard alias, so `m` here is just an example. Some packages do have near-universal conventions - `numpy` is almost always imported as `np` - and following those makes your code recognisable to other people.

### Importing only what you need

If you only want one or two things out of a module, import them by name. They then work on their own, without the module in front.

````python
# Import specific functions, using commas when more than one
from math import sqrt, asin

print(4 * asin(1 / sqrt(2)))
````

There is one more form. `from math import *` takes everything out of the module at once, and it comes with a warning.

````python
print(pow(2, 3))

from math import *

print(pow(2, 3))
````

The same line of code printed `8` and then `8.0`.

`pow` is a *built-in* function in Python, which is always available, and which returns a whole number. But the `math` module defines a function of the same name that always returns a decimal number. Importing everything from `math` replaced one with the other, and nothing in the code that calls `pow` tells you this has happened.

That is the problem with `from module import *`: it drops an unknown collection of names into your code, and any of them may overwrite something you were relying on. Best practice is to import the module, or import only the specific functions you need.

### The four forms

To summarise:

| Form | What it does | How you then refer to it |
|------|-------|-------|
| `import math` | Imports the whole module | `math.sqrt(2)` |
| `import math as m` | Imports the whole module under a shorter name | `m.sqrt(2)` |
| `from math import sqrt` | Imports one thing from the module | `sqrt(2)` |
| `from math import *` | Imports everything from the module | `sqrt(2)` |

Most of the time you will use one of the first two. The third is reasonable when you only want one or two things out of a module, and the fourth is best avoided.

### Where imports go

By convention, imports go at the very top of a file or Notebook, before any other code. This means anyone opening your file can see immediately what it depends on.

Where a file has several imports, they are conventionally grouped in this order, with a blank line between the groups:

1. Modules from the standard library
2. Packages installed from elsewhere
3. Modules from your own project

## Packages

A package is a collection of modules gathered into a directory. `numpy`, for example, is a package containing many modules, one of which is `numpy.random`.

In practice the words are used loosely, and you will hear "package", "module" and "library" used more or less interchangeably. Day to day, you'll just want to know whether the code is part of the standard library, or something you have to install.

The standard library comes with Python. Everything else has to be installed before you can import it, and if it is not installed, Python tells you so. Run the cell below.

````python
import definitely_not_a_real_package
````

`ModuleNotFoundError: No module named 'definitely_not_a_real_package'` means exactly what it says. Either the name is wrong, or the package is not installed in the environment you are using.

Packages that are not part of the standard library are published to the [Python Package Index](https://pypi.org), known as PyPI. There are several hundred thousand of them. Before writing anything substantial, check whether a well-established package already does it. The code will be better tested than yours, better documented, and other people will already know how it works.

### Installing packages with pip

`pip` is the tool that installs packages from PyPI. It is run from a terminal, not from a Notebook. In VS Code, the `Terminal` lives in the bottom panel, which you can show/hide using the buttons in the top-right of the window.

| Command | What it does |
|------|-------|
| `python -m pip install numpy` | Installs the `numpy` package |
| `python -m pip install numpy==2.1.0` | Installs one specific version |
| `python -m pip list` | Lists everything installed |
| `python -m pip show numpy` | Shows the version and location of one package |

Write `python -m pip` rather than plain `pip`: the extra characters guarantee the package goes into the same Python that will run your code.

This course's packages are already installed in your Codespace, so `numpy` imports without complaint.

````python
import numpy as np

print(np.random.random(5))
````

An environment is one installation of Python together with the packages installed into it, and a computer can have several. ## Writing Your Own Module

Any `.py` file you write is a module, and can be imported the same way as `math` or `numpy`.

Next to this Notebook are two of them, because modules rarely sit alone. The first holds functions about circles, and the second uses those functions to work with cylinders.

Open `circles.py` from the file explorer on the left and read it. Its contents are below.

```python
"""Functions for the geometry of circles."""

import math


def circle_area(radius):
    """Return the area of a circle of the given radius."""
    return math.pi * radius ** 2


def circle_circumference(radius):
    """Return the circumference of a circle of the given radius."""
    return 2 * math.pi * radius
```

Two notes:

* The file opens with a line of text in triple quotes, called a docstring, saying what the module is for. Each function has one too, in line with best practice.

* Notice that your own modules are free to import other Python modules, as we do with `math` here.

In the cell below, let's import this module. The name to import is the filename without the `.py`, and Python will find it because it sits in the same directory as this Notebook.

````python
import circles

print(circles.circle_area(1))
print(circles.circle_circumference(1))
````

### One module importing another

Now open the second file, `cylinders.py`:

```python
"""Functions for the geometry of cylinders."""

# This relies on circles.py being in the same directory as this file
from circles import circle_area


def cylinder_volume(radius, height):
    """Return the volume of a cylinder of the given radius and height."""
    return circle_area(radius) * height
```

Two notes:

* `cylinders.py` imports from `circles.py` with the usual syntax.

* Python looks for `circles` in the directory your code is being run from, so the two files have to sit together. Otherwise, you will get a `ModuleNotFoundError`.

Let's import the cylinders module now, using two different forms.

````python
import cylinders

print(cylinders.cylinder_volume(3, 5))
````

````python
from cylinders import cylinder_volume

print(cylinder_volume(3, 5))
````

`help()` works on your own modules as well. This is what the docstrings at the top of each function are for.

````python
help(cylinders)
````

### Imports and Notebooks

When working with Jupyter Notebooks, Python imports a module once and then keeps it in memory. If you edit `cylinders.py` now and run `import cylinders` again, **nothing will change**. Python sees that it has already imported `cylinders` and does not look at the file again.

The fix is to restart the kernel, using the `Restart` button in the toolbar at the top of the Notebook, and then run your cells again. The kernel is the Python process running behind the Notebook; restarting it clears every variable and every imported module, giving you a clean slate.

## Organising a Project

Notebooks and `.py` files have different jobs. A Notebook is for exploring and explaining: a person runs it top to bottom and reads the output. A `.py` file is for code you rely on: it gets imported, tested and reused. Code that has stopped changing and started being depended upon belongs in a `.py` file.

You have been working inside a project all morning. This Notebook is not sitting next to one, it is inside one:

```
9_packages_imports_projects/
├── README.md
├── packages_imports_projects.ipynb   <- you are here
├── circles.py
├── cylinders.py
├── dimensions.py
├── report_cylinders.py
├── data/
│   └── cylinders.csv
└── tests/
    └── test_cylinders.py
```

Open the file explorer on the left and check it against the tree above.

This is also why `import circles` worked earlier with no setting up. The Notebook sits at the top level of the project, so the directory Python searches is the directory the modules are in.

### One job per module

Look at the four `.py` files. Each one has a job you can state in a sentence:

* `circles.py` - the area and circumference of a circle
* `cylinders.py` - the volume of a cylinder, built on `circles.py`
* `dimensions.py` - reading the dimensions file
* `report_cylinders.py` - the script you actually run, which uses the other three

None of them is long, which is why they are split up in the first place. When something is wrong, you know which file to open.

`dimensions.py` is a module like any other, so you can import from it here too. Run the cell below.

````python
from dimensions import read_dimensions

cylinders = read_dimensions("data/cylinders.csv")

print(cylinders[0])
print(len(cylinders), "cylinders in the file")
````

### What each part is for

**The `.py` files at the top level** hold the code. Three of them are modules you import from, and `report_cylinders.py` is the script you run.

**`data/`** holds the dimensions file. Keeping data out of the code means you can point the project at a different file without editing anything.

**`tests/`** holds the tests, in their own directory so they stay out of the way of the code they test.

**`README.md`** says what the project is and how to run it. It's the first thing anyone opening your project reads, including you when you come back to it.

A bigger project will have more in it, but it will be arranged along the same lines, with the code, the data and the tests kept apart from each other.

### Running the tests

On Tuesday you ran tests inside a Notebook, with `%%ipytest` at the top of a cell. That kept a function and its tests side by side while you were writing both. `ipytest` is a wrapper around `pytest`, which is the tool doing the actual work.

In a project, the tests live in files of their own. Open `tests/test_cylinders.py` and you will recognise the shape: a function whose name begins with `test_`, and an `assert`.

VS Code will find and run them for you. Click the flask icon in the bar on the far left to open the Testing panel. You should see `test_cylinders.py` listed, with the tests inside it.

* The play button beside a test runs that one test. The play button at the top runs all of them.
* A test that passes gets a green tick, one that fails gets a red cross.
* Clicking a failed test shows you what it expected and what it got, and takes you to the line that failed.
