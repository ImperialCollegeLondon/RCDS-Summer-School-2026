## Generative AI

Generative AI is a technology trained on very large datasets. It learns patterns in sequences of tokens (words, characters, and code symbols) and predicts what token is likely to come next. This lets it produce fluent text and plausible code from a prompt.

That makes it useful for programming, but it also comes with important limitations. Some considerations are summarised below:

* **Generative AI can produce useful information on many topics**: Models are trained on broad datasets, so they can often give helpful answers quickly. However, generative AI can give entirely incorrect or fictitious arguments with convincing confidence.
* **Generative AI is fast**: It can generate drafts, examples, and code much faster than most people can write from scratch.
* **Generative AI does not think like a human**: It has no understanding or consciousness. It generates output from statistical patterns, so it may make cognitive leaps people would not, and also mistakes people would not.
* **Generative AI is not always correct**: Generative AI models can produce plausible-looking code with subtle bugs, missing edge cases, or incorrect assumptions.
* **Generative AI has limited working memory**: It relies on a finite "context window". In long tasks, it can lose track of earlier details and become inconsistent.

The key consequence is that generative AI can be a powerful programming partner, but it needs active human oversight. You should critically evaluate outputs and verify correctness.

Tasks that generative AI tends to perform poorly at include:

* Keeping track of the overall project goal over long periods
* Relating code decisions to a specific research or real-world context
* Choosing which behaviours and edge cases matter most in your context
* Judging when to refactor or reorganise code
* Planning what to do next in larger projects
* Retaining long-term context across many iterations
* Simplifying code by deleting or combining parts (it often adds more than it removes)

These are responsibilities that should stay with you. For focused tasks, such as drafting a function or making a small feature change, AI can accelerate your work, but only with careful steering and review.

### Environmental Impact

### Cost

Generative AI models are expensive to train and run. Users are often billed by token usage, so cost increases with longer prompts, longer outputs, and additional model reasoning.

### Judicious Use

Environmental and cost concerns both encourage using generative AI judiciously: use it where it has clear value, and use it efficiently.

## Which Generative AI Tool to Use When Programming

Many of you will have used generative AI tools such as ChatGPT through a web interface. This can be a good way to use generative AI for some tasks, but it has downsides when working on a programming project.

For coding work, it is often better to use a tool that integrates with your code editor, such as GitHub Copilot, which we will use in this course. This has several benefits:

* The AI is in the same interface as your code, so you do not need to keep switching windows.
* The AI can see your code and project structure, so it can provide more relevant suggestions. It also reduces copy-pasting of existing code into the AI.
* The AI can propose changes directly in your file, which you can review and accept. This reduces effort and copy-paste errors.
* In supported setups, the AI can run your code or tests and use the results to improve its suggestions and help with debugging.
* Coding assistants such as GitHub Copilot are typically tuned for programming tasks, so they often provide stronger coding support than general-purpose models at a similar cost or complexity.

#### Agents

Copilot has three different modes that can be selected through a button at the bottom of the chat interface:
* **Agent**: Ask Copilot to carry out a task or series of tasks. Copilot may propose code changes, run tools, create, delete or rename files, etc. Often, Copilot will ask for permission before carrying out tasks such as running terminal commands or other potentially destructive actions.
* **Ask**: Ask Copilot questions (about your code or otherwise). Copilot will not propose changes to your code directly in the editor.
* **Plan**: Ask Copilot to plan a task or series of tasks. It will propose a plan, which you can accept or modify. In Plan mode, Copilot will not make changes to your code. Once you have agreed on a plan, you can switch to Agent mode to have Copilot carry it out.

It is also possible to define custom agents with custom instructions, but that is beyond the scope of this introduction.

#### Context Window

An agent's working memory is defined by its context window. This is its memory of the current conversation, the files its looked at, any context and instructions. The fullness of the current context window is displayed at the bottom of the chat interface as a ring that gets more full as the context window fills. If the context window fills completely, the agent will internally produce a summary of its memory, effectively compressing the information it has seen into a smaller representation. This summary will be used in place of the original context, which can lead to the agent forgetting details.

When you prompt the agent, it will examine the contents of its context window and use that information to generate a response. As the window fills, this will increase the token usage of each prompt and the latency of the response.

#### Starting a New Session

It's often a good idea to start a new session when changing tasks due to the behaviour of the context window. This can minimise token usage and improve response times. However, this will also clear the context window, so you will need to provide any relevant context again. You can do this by clicking the `+` button at the top of the chat interface.

### Prompting Tips

You can ask an agent to do almost anything, but some tips will help you get the best results:

* **Be specific**: If you want a function to do something, describe the inputs, outputs, and behaviour you expect.
* **Set precedent**: If adding to an existing codebase, show the agent examples of the style and structure you want it to follow already present in the codebase.
* **Plan**: If you are about to perform a complex task, ask the agent to plan it first. This can help you avoid mistakes and ensure the agent understands your goals.
* **Muse**: If you have thoughts on how something might work, but aren't sure, present your ideas, even if they are incomplete or uncertain. Ask for feedback and suggestions. This can sometimes cause the agent to generate ideas you hadn't considered, and can help you clarify your own thinking.
* **Query**: If you don't understand something, ask the agent to explain it. Sometimes you'll uncover a flaw in what the agent has done. Other times, you might learn something new.
* **Iterate**: If the proposed solution isn't right, describe any shortcomings and try again.
* **Trim**: Agents are more prone to add code than to remove or consolidate it. If you think a piece of code has been made redundant, ask the agent if it is now redundant and can be removed. If you think functionality has been duplicated, ask the agent if it can be consolidated.
* **Test**: Reviewing the code is a good start, but you should also run the code manually, or with a testing framework, to ensure it behaves as expected. You can also ask the agent to generate or execute tests.
