## For Loops

In programming it is common to want to perform the same operation repeatedly and, in Python, `for` loops allow us to do this. A `for` loop allows us to iterate over a sequence of values, performing the same operation for each value in the sequence. If we want to iterate over a range of numbers, the syntax is as follows:

````python
for i in range(5):
    x = i * 2
    print(x)

print("Done!")
````

The word `for` tells Python we want to create a loop. Next we write the name of a variable (in this case `i`) which is known as the loop variable. This variable will take on each value in the sequence, one at a time, as the loop iterates. The `in` keyword is used to specify the sequence we want to iterate over, which in this case is generated by the `range()` function. The `range()` function generates an object which represents a sequence of numbers. In this case, `range(5)` generates the numbers 0, 1, 2, 3 and 4 and these values are given to `i` in subsequent iterations of the loop. Next, we write a colon (`:`) to indicate that the next indented block of code is the body of the loop. The body of loop is indicated through indentation (often created by pressing the tab key). The body of the loop is executed once for each value in the sequence. In this case, the body of the loop multiplies `i` by 2 and assigns the result to a variable `x`, which is then printed. After the loop has finished iterating over all values in the sequence, execution moves to the unindented line of code after the loop, which prints "Done!".

### The Accumulator Pattern

There are many useful things that can be done with a `for` loop, but one of the most common is to use it to calculate a total or sum, using an approach known as the accumulator pattern. The accumulator pattern involves creating a variable to hold the total, which is initialised to zero before the loop starts. Then, inside the loop, we add to this variable in each iteration. After the loop has finished, the variable will contain the total sum of all values added to it. In the example below, we use the accumulator pattern to calculate the sum of each number between 1 and 10 (inclusive).

Remember, the `range()` function generates numbers up to, but not including, the second argument, so we need to use `range(11)` to include 10 in the range.

Remember `x += i` is shorthand for `x = x + i`, which adds the value of `i` to `x` and assigns the result back to `x`.

````python
x = 0

for i in range(11):
    x += i

print(x)
````

### Controlling the range

We can ask the `range` to generate different patterns of numbers by providing it with different numbers of arguments:

* 1 argument: `range(n)` generates numbers from 0 to n-1 (inclusive)
* 2 arguments: `range(start, stop)` generates numbers from start to stop-1 (inclusive)
* 3 arguments: `range(start, stop, step)` generates numbers from start to stop-1 (inclusive), incrementing by step

For example:

````python
print("One argument: range(n) generates numbers from 0 to n-1 (inclusive)")

for i in range(5):
    print(i)

print("Two arguments: range(start, stop) generates numbers from start to stop-1 (inclusive)")

for i in range(2, 5):
    print(i)

print("Three arguments: range(start, stop, step) generates numbers from start to stop-1 (inclusive), incrementing by step")

for i in range(2, 10, 2):
    print(i)
````

### Nesting Constructs

In Python, it is completely acceptable to nest constructs, such as `for` loops and `if` statements, inside one another. When we do this, the interior of the second construct is indented twice. For example:

````python
# Print every other number between 5 and 100 smaller than 20
for i in range(5, 100, 2):
    if i < 20:
        print(i)
````

## Testing

Testing is a way to check your code behaves as expected. Tests are small pieces of code which are stored alongside the main code of your project. They will typically call functions from your main code with different inputs and check that the output is what you expect. If the output is not what you expect, the test will fail and you will be alerted to this. It is common that a project will have many tests, and the production of tests is an important part of the software development process. Some advantages of tests include:

* A failing test will alert you to a problem in your code, which can save you time and effort in debugging.
* Writing tests alongside your code encourages you to think about the expected behaviour of your code, which can help you to write better code.
* Tests can be used to check that changes to your code do not break existing functionality. This is particularly important in larger projects where multiple people may be working on the same codebase.
* Tests can increase confidence in your code, which can be particularly important when you are working on a project with other people or if you're producing code that will be used by others.

Tests can be run manually, or automatically as part of the software engineering process.

### Testing Frameworks

Testing frameworks are tools which can make running tests easier. They allow you to organise tests into groups and provide the ability to run subsets of tests or all tests at once and report summaries of results and details of failures. They also provide a standardised way to write tests, which can make it easier to read and understand tests written by other people. There are many testing frameworks available for Python, but one of the most popular is [`pytest`](https://docs.pytest.org/en/stable/). In this course we will use `pytest` to run our tests.

### A Simple Example

Let's imagine we have just written a function which adds two numbers together:

````python
def addition(a, b):
    return a + b
````

Using `pytest` we might write some tests for this function:

````python
%%ipytest

def test_addition_both_zero():
    assert addition(0, 0) == 0

def test_addition_positive_numbers():
    assert addition(2, 3) == 5

def test_addition_negative_numbers():
    assert addition(-2, -3) == -5
````

In the cell above, the command `%%ipytest` tells the notebook to run the code in the cell as a collection of `pytest` tests.

Each function is a test and should begin with the word `test_` so `pytest` recognises it as a test. Each test will be run independently and will succeed if their is no error, and fail if there is an exception.

A common way to cause the test to fail if there is a problem is to use the `assert` statement. This is followed by a condition which should be true if the code is behaving as expected. If the condition is false, the test will fail and `pytest` will report this.

The code cell below contains a test which is designed to fail. Run the cell and see what happens. You should see a report of the test results, including details of the failure.

````python
%%ipytest

def test_passing():
    assert addition(1, 1) == 2

def test_deliberately_failing():
    assert addition(2, 2) == 5, "This was destined to fail"  # This test is expected to fail
````

In the example above, we also followed the `assert` statement of the failing check with a comma, then a message in quotation marks. This message will be displayed in the test report if the test fails, and can be used to provide additional information about the failure.

### Creating Tests

When creating tests, it is usually a good idea to test one scenario in each test. As in the example above, this means a single function might have several tests. This is useful because we can get a hint about the nature of the problem by which tests have failed and which have passed.

It is a good idea to test a variety of scenarios, including edge cases. For example, if we were testing a function which calculates the square root of a number, we might want to test the following scenarios:

* Zero (0)
* A positive number with an integer square root (e.g. 4)
* A positive number with a non-integer square root (e.g. 2)
* A non-integer number (e.g. 2.25)
* A negative number (e.g. -4)
* A non-numeric value (e.g. "hello")

This not only tests a variety of scenarios, but also forces us to think about what the expected behaviour of our code should be in each of these scenarios. This can help us to write better code and avoid bugs.

### Floating Point Error and Tests

Like all programming languages, Python cannot store floating point numbers to infinite precision. Instead, the results of floating point calculations are the nearest value Python can represent. This can lead to small rounding errors, known as floating point error. Consider the code below:

````python
# Analytically, x and y are the same
x = 1 - 0.99999
y = 0.00001

# In Python floating point error means they are not the same
print(x)
print(y)

# This means comparing one to the other returns False
print(x == y)
````

This causes problems with tests. The following test will fail, even though the code we wrote was mathematically correct.

````python
%%ipytest

def subtract(x, y):
    return x - y

def test_subtract():
    # This test will fail due to floating point error
    assert subtract(1, 0.99999) == 0.00001
````

Fortunately, the `pytest` package provides the `approx` function, which can be applied to one of the floating point numbers being compared in an assertion. This will allow the test to pass even if there are minor floating point inaccuracies.

````python
%%ipytest
import pytest # Notice this line needs to be present for the pytest.approx function to work correctly

def subtract(x, y):
    return x - y

def test_subtract():
    # This test will fail due to floating point error
    assert 0.00001 == pytest.approx(subtract(1, 0.99999))
````
