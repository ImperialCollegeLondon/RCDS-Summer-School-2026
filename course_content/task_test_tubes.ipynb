## Background

You have recently inherited a test tube factory and you need to order glass for a new consignment.

All test tubes produced by your factory conform to the following design:

The tube is formed of glass of a thickness $t$ which is constant over the whole surface. 

It has a hemispherical base with an outer radius of $r$ and, thus, an inner radius of $r − t$

Above the hemisphere is a cylinder of height $h$ and outer radius $r$.
